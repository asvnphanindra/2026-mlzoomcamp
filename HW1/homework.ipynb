## Q6. Median value of horsepower

1. Find the median of `horsepower`
2. Find the most frequent value (mode)
3. Fill missing values with the mode
4. Compute the median again — did it change?

```python
median_before = df["horsepower"].median()
mode_hp = df["horsepower"].mode()[0]

horsepower_filled = df["horsepower"].fillna(mode_hp)
median_after = horsepower_filled.median()

print(f"Median before fill: {median_before}")
print(f"Most frequent value: {mode_hp}")
print(f"Median after fill:  {median_after}")

if median_after > median_before:
    print("Changed: Yes, it increased")
elif median_after < median_before:
    print("Changed: Yes, it decreased")
else:
    print("Changed: No")
```

**Answer Q6:** `Yes, it decreased`

## Q7. Sum of weights

Linear-regression style calculation with NumPy on Asia cars.

```python
# 1-3. Asia cars, vehicle_weight + model_year, first 7 rows
X = (
    df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]]
    .iloc[:7]
    .to_numpy()
)

# 5-6. XTX and its inverse
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

# 7-8. Compute w
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y

# 9. Sum of elements of w
print("X:")
print(X)
print("w:", w)
print("sum(w):", w.sum())
```

**Answer Q7:** `0.369`
